```python

        # Create a logger
from xact.log.config import log_manager


logger1 = log_manager.init("Logger1")
    
# Create another logger
logger2 = log_manager.init("Logger2", level="debug")

logger1.info("This is a info message from Logger1.")
logger2.debug("This is a debug message from Logger2.")
logger1.info("Logger1 with new format.")
logger2.debug("Logger2 with new format.")

log_manager.config(level="WARNING")
logger1.info("This will not be shown because the level is now WARNING.")
logger2.warning("This is a warning message from Logger2.")

# Setup a custom logging level
log_manager.setup_custom_level("verbose", 50)
logger1.verbose("This is a verbose message from Logger1.")
```
