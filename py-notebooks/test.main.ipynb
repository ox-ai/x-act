```python
import re
import json

def preprocess_json(json_str: str):
    """
    Preprocess a string to remove comments and ensure JSON compatibility.

    Args:
        json_str (str): The JSON string with possible comments.

    Returns:
        str: A cleaned JSON string.
    """
    # Remove single-line comments (`#` and `//`)
    json_str = re.sub(r"(?<!:)#.*", "", json_str)  # Python-style comments
    json_str = re.sub(r"//.*", "", json_str)       # C++/JavaScript-style comments
    
    # Remove trailing commas that may make the JSON invalid
    json_str = re.sub(r",\s*([\]}])", r"\1", json_str)

    return json_str


def extract_json_objects(text: str):
    """
    Extract JSON objects from a string, including nested JSON.

    Args:
        text (str): The input string containing JSON-like objects.

    Returns:
        list: A list of valid JSON objects as dictionaries.
    """
    stack = []
    json_objects = []
    start_idx = None

    for i, char in enumerate(text):
        if char == '{':
            if not stack:
                start_idx = i
            stack.append(char)
        elif char == '}':
            if stack:
                stack.pop()
                if not stack:  # Complete JSON object
                    json_str = text[start_idx:i+1]
                    try:
                        # Preprocess and parse JSON
                        cleaned_json = preprocess_json(json_str)
                        json_obj = json.loads(cleaned_json)
                        json_objects.append(json_obj)
                    except json.JSONDecodeError:
                        pass  # Skip invalid JSON

    return json_objects


# Example Usage
llm_output = """
Here's an updated prompt based on historical data:

{"updated_prompt":"get summary of plan for 25th dec 2024 to 26th dec 2024",
"day":["December", "2024"],  # vague, but covers the date range
"time":"December holidays"}}

Additional context:
{"context":{"level1":{"level2":{"key":"value"}}, "array":[1, 2, 3]},  // nested JSON example
"note":"sample comment"}
"""

# Extract JSON from the example LLM output
extracted_jsons = extract_json_objects(res)

print(extracted_jsons[0])
```

```python
from datetime import datetime
import re

def convert_time_string_to_datetime(time_string:str):
    """
    Converts a time string in various formats (e.g., "9am", "9:00", "09:00 AM", "17:30") to a datetime.time object.

    Args:
        time_string (str): The time string to convert.a

    Returns:
        datetime.time or None: A datetime.time object representing the parsed time, or None if the input is invalid.
    """
    formats_to_try = [
        "%I",
        "%I%p",        # 9am, 10pm (12-hour format with AM/PM)
        "%I:%M%p",     # 9:00am, 10:30pm
        "%H:%M",       # 09:00, 17:30 (24-hour format)
        "%H:%M:%S",   # 09:00:00, 17:30:00
        "%I:%M:%S%p" #9:00:00am, 10:30:00pm
    ]
    time_string = re.sub(r"\s+", "", time_string.strip())

    for fmt in formats_to_try:
        try:

            time_obj = datetime.strptime(time_string, fmt).time()
            return time_obj
        except ValueError:
            pass  # Try the next format

    return None  # If no format matches

# Example usage:
times_to_test = ["9am", "9:00", "09:00 AM", "17:30", "9:00:00am","17:30:00","9:00:00"]
for time_str in times_to_test:
    time_obj = convert_time_string_to_datetime(time_str)
    if time_obj:
        print(f"'{time_str}' converts to: {time_obj}")
    else:
        print(f"Could not parse time string: '{time_str}'")

# more test cases
test_cases = [
    ("12pm", "12:00:00"),
    ("12am", "00:00:00"),
    ("1:00pm", "13:00:00"),
    ("01:00 PM", "13:00:00"),
    ("01:00:00 PM", "13:00:00"),
    ("13:00", "13:00:00"),
    ("13:00:00", "13:00:00"),
    ("1:00", "01:00:00"),
    ("01:00", "01:00:00"),
    ("9", "09:00:00"), # fails because it needs am/pm
    ("9am", "09:00:00"),
    ("9:00", "09:00:00"),
    ("09:00 AM", "09:00:00"),
    ("17:30", "17:30:00")
]

for time_str, expected in test_cases:
    
    result = convert_time_string_to_datetime(time_str)
    print(time_str,"===",result)

    result_str = result.strftime("%H:%M:%S")
    assert result_str == expected, f"Test failed for '{time_str}': Expected '{expected}', got '{result_str}'"
    print(f"'{time_str}' passes")


print("All tests passed!")
```

```python
from datetime import datetime
import re

def convert_date_string_to_datetime(date_string: str, default_year: int = datetime.now().year):
    """
    Converts a date string in various formats (e.g., "25th Dec", "24 Feb", "25 Dec 2023") to a datetime.date object.

    Args:
        date_string (str): The date string to convert.
        default_year (int): The default year to use if no year is provided.

    Returns:
        datetime.date or None: A datetime.date object representing the parsed date, or None if the input is invalid.
    """
    # Remove ordinal suffixes (st, nd, rd, th) from date
    date_string = re.sub(r'(\d+)(st|nd|rd|th)', r'\1', date_string.strip().lower())

    # Predefined formats to try
    formats_to_try = [
        "%d %b",          # 25 Dec
        "%d %B",          # 25 December
        "%d %b %Y",       # 25 Dec 2023
        "%d %B %Y",       # 25 December 2023
        "%d-%b",          # 25-Dec
        "%d-%b-%Y",       # 25-Dec-2023
        "%d-%m",          # 25-12
        "%d-%m-%Y",       # 25-12-2023
    ]

    for fmt in formats_to_try:
        try:
            # Add default year if not present
            # if "%Y" not in fmt:
            #     date_string += f" {default_year}"
   
            # Parse the date
            date_obj = datetime.strptime(date_string, fmt).date()
            return date_obj
        except ValueError:
            pass  # Try the next format

    return None  # If no format matches


# Example usage:
dates_to_test = [
    "25th Dec", "24 Feb", "25 Dec 2023", "1st January", "31-Dec", "25-12", "25-12-2023", "15th Aug 1947"
]

for date_str in dates_to_test:
    date_obj = convert_date_string_to_datetime(date_str)
    if date_obj:
        print(f"'{date_str}' converts to: {date_obj}")
    else:
        print(f"Could not parse date string: '{date_str}'")

# Test cases
test_cases = [
    ("25th Dec", "2024-12-25"),
    ("24 Feb", "2024-02-24"),
    ("25 Dec 2023", "2023-12-25"),
    ("1st January", "2024-01-01"),
    ("31-Dec", "2024-12-31"),
    ("25-12", "2024-12-25"),
    ("25-12-2023", "2023-12-25"),
    ("15th Aug 1947", "1947-08-15")
]

for date_str, expected in test_cases:
    result = convert_date_string_to_datetime(date_str)
    result_str = result.strftime("%Y-%m-%d")
    assert result_str == expected, f"Test failed for '{date_str}': Expected '{expected}', got '{result_str}'"
    print(f"'{date_str}' passes")

print("All tests passed!")
```

```python
import re
from datetime import datetime, timedelta
from dateutil.relativedelta import relativedelta

# Define patterns and corresponding operations
TIME_REFERENCE_PATTERNS = {
    r"today": lambda now, _: now,
    r"tomorrow": lambda now, _: now + timedelta(days=1),
    r"yesterday": lambda now, _: now - timedelta(days=1),
    r"in (?P<number>\d+) days?": lambda now, match: now + timedelta(days=int(match.group("number"))),
    r"after (?P<number>\d+) days?": lambda now, match: now + timedelta(days=int(match.group("number"))),
    r"last year": lambda now, _: now - relativedelta(years=1),
    r"after (?P<number>\d+) months?": lambda now, match: now + relativedelta(months=int(match.group("number"))),
    r"last month": lambda now, _: now - relativedelta(months=1),
    r"after (?P<number>\d+) hours?": lambda now, match: now + timedelta(hours=int(match.group("number"))),
    r"after (?P<number>\d+) minutes?": lambda now, match: now + timedelta(minutes=int(match.group("number"))),
    r"on (?P<day>\d{1,2})(?:st|nd|rd|th)? (?P<month>[a-zA-Z]+) of (?P<year>\d{4})": lambda now, match: datetime.strptime(f"{match.group('day')} {match.group('month')} {match.group('year')}", "%d %B %Y"),
    r"on (?P<day>\d{1,2})(?:st|nd|rd|th)? (?P<month>[a-zA-Z]+)": lambda now, match: datetime.strptime(f"{match.group('day')} {match.group('month')} {now.year}", "%d %B %Y"),
    r"(?P<number>\d{1,2}):(?P<minute>\d{2}) ?(?P<meridian>am|pm)": lambda now, match: now.replace(hour=int(match.group("number")) % 12 + (12 if match.group("meridian").lower() == "pm" else 0), minute=int(match.group("minute"))),
}


def parse_date_time(input_string: str) -> datetime:
    now = datetime.now()
    input_string = input_string.lower().strip()

    for pattern, operation in TIME_REFERENCE_PATTERNS.items():
        match = re.search(pattern, input_string)
        if match:
            return operation(now, match)

    raise ValueError(f"Could not parse date/time reference: {input_string}")


# Example Usage
examples = [
    "Today",
    "Tomorrow",
    "Yesterday",
    "In three days",
    "Last year",
    "After 8 months",
    "Last month",
    "After 1 hour",
    "After 20 minutes",
    "Tomorrow 9am",
    "On 23 Dec",
    "On 25th Feb of 2023",
]

for example in examples:
    try:
        result = parse_date_time(example)
        print(f"Input: {example} -> Parsed Date/Time: {result}")
    except ValueError as e:
        print(e)
```

```python
import re
from datetime import datetime, timedelta
from dateutil.relativedelta import relativedelta

# Define mapping for time references
TIME_REFERENCE_MAPPING = {
    "today": {"day": 0},
    "tomorrow": {"day": 1},
    "yesterday": {"day": -1},
    r"in (?P<number>\d+) days?": {"day": "\g<number>"},
    r"after (?P<number>\d+) days?": {"day": "\g<number>"},
    "last year": {"year": -1},
    r"after (?P<number>\d+) months?": {"month": "\g<number>"},
    "last month": {"month": -1},
    r"after (?P<number>\d+) hours?": {"hour": "\g<number>"},
    r"after (?P<number>\d+) minutes?": {"minute": "\g<number>"},
    r"on (?P<day>\d{1,2})(?:st|nd|rd|th)? (?P<month>[a-zA-Z]+) of (?P<year>\d{4})": {"date": "\g<day>-\g<month>-\g<year>"},
    r"on (?P<day>\d{1,2})(?:st|nd|rd|th)? (?P<month>[a-zA-Z]+)": {"date": "\g<day>-\g<month>"},
    r"(?P<hour>\d{1,2}):(?P<minute>\d{2}) ?(?P<meridian>am|pm)": {"time": "\g<hour>:\g<minute>:00 \g<meridian>"},
}


def apply_mapping(now: datetime, mapping: dict) -> datetime:
    """Apply a mapping of date/time adjustments to the current datetime."""
    for key, value in mapping.items():
        if key == "day":
            now += timedelta(days=int(value))
        elif key == "month":
            now += relativedelta(months=int(value))
        elif key == "year":
            now += relativedelta(years=int(value))
        elif key == "hour":
            now += timedelta(hours=int(value))
        elif key == "minute":
            now += timedelta(minutes=int(value))
        elif key == "date":
            now = datetime.strptime(value, "%d-%b-%Y") if "-" in value else datetime.strptime(value + f"-{now.year}", "%d-%b-%Y")
        elif key == "time":
            time_parts = datetime.strptime(value, "%I:%M:%S %p")
            now = now.replace(hour=time_parts.hour, minute=time_parts.minute, second=time_parts.second)
    return now


def parse_date_time(input_string: str) -> datetime:
    now = datetime.now()
    input_string = input_string.lower().strip()

    for pattern, mapping in TIME_REFERENCE_MAPPING.items():
        match = re.search(pattern, input_string)
        if match:
            # Replace placeholders in mapping with actual matched values
            resolved_mapping = {
                k: (v if not isinstance(v, str) else re.sub(pattern, v, input_string))
                for k, v in mapping.items()
            }
            return apply_mapping(now, resolved_mapping)

    raise ValueError(f"Could not parse date/time reference: {input_string}")


# Example Usage
examples = [
    "Today",
    "Tomorrow",
    "Yesterday",
    "In three days",
    "Last year",
    "After 8 months",
    "Last month",
    "After 1 hour",
    "After 20 minutes",
    "Tomorrow 9am",
    "On 23 Dec",
    "On 25th Feb of 2023",
]

for example in examples:
    try:
        result = parse_date_time(example)
        print(f"Input: {example} -> Parsed Date/Time: {result}")
    except ValueError as e:
        print(e)
```

```python
from datetime import datetime, timedelta
from dateutil.relativedelta import relativedelta

def apply_mapping(now: datetime, mapping: dict) -> datetime:
    """Apply a mapping of date/time adjustments to the current datetime."""

    adjustments = {
        "day": lambda value: timedelta(days=int(value)),
        "month": lambda value: relativedelta(months=int(value)),
        "year": lambda value: relativedelta(years=int(value)),
        "hour": lambda value: timedelta(hours=int(value)),
        "minute": lambda value: timedelta(minutes=int(value)),
        "date": lambda value: datetime.strptime(value, "%d-%b-%Y") if "-" in value else datetime.strptime(value + f"-{now.year}", "%d-%b-%Y"),
        "time": lambda value: now.replace(hour=datetime.strptime(value, "%I:%M:%S %p").hour, minute=datetime.strptime(value, "%I:%M:%S %p").minute, second=datetime.strptime(value, "%I:%M:%S %p").second)
    }

    for key, value in mapping.items():
        if key in adjustments:
            if key in ("date","time"):
                now = adjustments[key](value)
            else:
                now += adjustments[key](value)

    return now

# Example Usage
now = datetime(2024, 10, 26, 10, 30, 0)
mapping1 = {"day": 2, "month": -1, "time": "09:00:00 AM"}
mapping2 = {"date": "20-Nov", "time": "05:00:00 PM"}
mapping3 = {"date": "20-Nov-2025", "time": "05:00:00 PM"}

updated_now1 = apply_mapping(now, mapping1)
print(f"Updated datetime 1: {updated_now1}")

updated_now2 = apply_mapping(now, mapping2)
print(f"Updated datetime 2: {updated_now2}")

updated_now3 = apply_mapping(now, mapping3)
print(f"Updated datetime 3: {updated_now3}")

now = datetime(2024, 1, 31, 10, 30, 0)
mapping4 = {"month": 1}
updated_now4 = apply_mapping(now, mapping4)
print(f"Updated datetime 4: {updated_now4}")
```
