```python
import asyncio

# Create an instance of BashTool
bash_tool = bash.BashTool()

async def main():
    # Start the BashTool session and run a command
    print("Starting BashTool and running the 'echo' command:")
    result = await bash_tool(command="echo Hello, BashTool!")
    print("Output:", result.output)  # Should display 'Hello, BashTool!'
    print("Error:", result.error)  # Should be empty if there's no error

    # Run another command in the same session
    print("\nRunning 'ls' command to list files in the current directory:")
    result = await bash_tool(command="ls")
    print("Output:", result.output)
    print("Error:", result.error)

    # Restart the BashTool session and run a command in the new session
    print("\nRestarting BashTool session and running 'pwd' command:")
    result = await bash_tool(command="ls", restart=True)
    print("Output:", result.output)
    print("Error:", result.error)

# Run the async main function
# Run the async main function in a Jupyter Notebook
await main()

```
