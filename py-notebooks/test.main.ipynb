```python
import json
from typing import List
from xact.flow.tool.tool import Tool

class FlowNode:
    def __init__(self, name:str):
        self.name = name

    def __str__(self):
        return f"{self.name}"

class FlowGraph:
    def __init__(self):
        self.graph = {}
        self.nodes = {}

    def __str__(self):
        return json.dumps(self.graph, indent=4)

    def add_node(self, func):
        
        toolx = Tool(func=func)
        if not self.nodes:
            self._start_node = toolx.name
        self.nodes[toolx.name] = toolx
        self.graph[toolx.name] = []

    def add_edge(self, source_node, nodex):
        
        self.graph[source_node].append(nodex)



    def add_edges(self,source_node, nodex_list:List[FlowNode]):
        for nodex in nodex_list:
            if nodex not in self.graph[source_node]:
                self.graph[source_node].append(nodex)


    def run(self):
        current_node = self._start_node
        while True:
            print(f"running : {current_node}")
            result = self.nodes[current_node].run()
            print(f"result : {result}")
            print(f"next nodes : {self.graph[current_node]}")
            next_node = input("next node : ")
            current_node = next_node
            if not current_node or current_node == "./end":
                break
        
```

```python
def shutdown():
    return "shutting down"


def open_apps(name: str="app"):
    """open apps from the given name"""
    return f"{name} app opened"

def update ():
    return "updated"

def restart():
    return "restarting"





gp = FlowGraph()


gp.add_node(open_apps)
gp.add_node(update)
gp.add_node(shutdown)
gp.add_node(restart)

gp.add_edge("open_apps", ["update","restart","shutdown"])
gp.add_edges("update",["restart","shutdown"])
gp.add_edges("restart",["shutdown"])



print(gp)
```
